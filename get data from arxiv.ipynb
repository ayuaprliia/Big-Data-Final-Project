```python
import arxiv
import pandas as pd
import time
from datetime import datetime
```

```python
# memilih sub category
subcategories = [
    'cs.AI', 'cs.CE', 'cs.CL', 'cs.CR', 'cs.CV', 'cs.DB', 'cs.IR',
    'cs.PL', 'cs.SE', 'cs.GT', 'cs.MA',
    'eess.AS', 'eess.IV', 'eess.SP', 'eess.SY'
]

# range tahun
start_year = 2020
end_year = 2025
```

```python
# Fungsi pembuat rentang bulan
def daterange(start_date, end_date):
    current = start_date
    while current < end_date:
        yield current
        if current.month == 12:
            current = current.replace(year=current.year + 1, month=1)
        else:
            current = current.replace(month=current.month + 1)
```

```python
# Function untuk fetch artikel dari arxiv berdasarkan tanggal 
def fetch_papers_by_month(category, start_year, end_year, max_per_month=300):
    papers = []
    start_date = datetime(start_year, 1, 1)
    end_date = datetime(end_year + 1, 1, 1)

    client = arxiv.Client(
        page_size=100,
        delay_seconds=3.0,   
        num_retries=3       
    )

    for single_month in daterange(start_date, end_date):
        month_start = single_month
        if single_month.month == 12:
            month_end = single_month.replace(year=single_month.year + 1, month=1)
        else:
            month_end = single_month.replace(month=single_month.month + 1)

        query = f"cat:{category} AND submittedDate:[{month_start.strftime('%Y%m%d0000')} TO {month_end.strftime('%Y%m%d0000')}]"

        search = arxiv.Search(
            query=query,
            max_results=max_per_month,
            sort_by=arxiv.SortCriterion.SubmittedDate,
            sort_order=arxiv.SortOrder.Descending
        )

        count = 0
        try:
            for result in client.results(search):
                papers.append({
                    'id': result.entry_id,
                    'title': result.title,
                    'authors': ', '.join([author.name for author in result.authors]),
                    'summary': result.summary,
                    'published': result.published.strftime('%Y-%m-%d'),
                    'updated': result.updated.strftime('%Y-%m-%d'),
                    'primary_category': result.primary_category,
                    'categories': ', '.join(result.categories),
                    'pdf_url': result.pdf_url
                })
                count += 1

            print(f"{category} | {month_start.strftime('%Y-%m')} : {count} makalah")

        except Exception as e:
            print(f"Gagal mengambil data untuk {category} {month_start.strftime('%Y-%m')} karena: {e}")
            continue

    return papers
```

```python
# proses fetching data
all_papers = []

for category in subcategories:
    print(f"\n Mengambil data untuk kategori: {category}")
    papers = fetch_papers_by_month(category, start_year, end_year, max_per_month=300)
    all_papers.extend(papers)
    print(f"Total artikel untuk {category}: {len(papers)}\n")

# save to csv
df = pd.DataFrame(all_papers)
df.to_csv('arxiv_papers_2020_2025.csv', index=False)
print("Data berhasil disimpan ke 'arxiv_papers_2020_2025.csv'")
```
